```python
#Import necessary libraries
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM,GRU,SimpleRNN,Embedding,Dense
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
#Load dataset
vocabulary=10000
(x_train,y_train),(x_test,y_test)=imdb.load_data()
sequence_length=200
x_train_padded=pad_sequences(x_train,maxlen=sequence_length)
x_test_padded=pad_sequences(x_test,maxlen=sequence_length)
```

```python
#Train the model
model = Sequential([
    Embedding(input_dim=10000,output_dim=128,input_length=sequence_length),
    SimpleRNN(128),
    Dense(1)
])
model.compile(optimizer='sgd',loss='binary_crossentropy',metrics=['accuracy'])
history=model.fit(x_train_padded,y_train,epochs=5,batch_size=64,validation_split=0.2)
loss,accuracy=model.evaluate(x_test_padded,y_test)
```
