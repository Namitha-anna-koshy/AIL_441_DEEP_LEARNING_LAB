```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
def image_enhancement(image_path):
    img=cv2.imread(image_path)
    gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
    hist_eq=cv2.equalizeHist(gray)
    _,binary=cv2.threshold(gray,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)
    erosion=cv2.erode(binary,np.ones((5,5),np.uint8),iterations=1)
    images=[cv2.cvtColor(img,cv2.COLOR_BGR2RGB),gray,hist_eq,binary,erosion]
    titles=['Orginal','Gray Scale','Histogram Equalizer','Binary','Erosion']
    plt.figure(figsize=(10,6))
    for i in range(5):
        plt.subplot(2,3,i+1)
        cmap='gray' if len(images[i].shape)==2 else None
        plt.imshow(images[i],cmap)
        plt.title(titles[i])
        plt.axis('off')
    plt.subplot(2,3,6)
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```
